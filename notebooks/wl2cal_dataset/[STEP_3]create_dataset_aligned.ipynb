# First step, extract all drawings (up to 2023)

```python
# open given source archive, print its root directory


# specifying the zip file name
original_zip = "../datasets/ImageTranslation-2012_2017/_zipfiles/drawing/validated_drawings.zip"
output_dir = "../datasets/ImageTranslationDEC2023/source_DR-UPTO2023"
years_images_to_extract  = [2018,2019,2020]
# years_images_to_extract  = [2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023]

content_list = []
content_dict = {}

# opening the zip file in READ mode
with ZipFile(original_zip, 'r') as zip:
    
    content_list = zip.namelist()

    # filter only the png and jpg files
    content_list = [x for x in content_list if x.endswith(".png") or x.endswith(".jpg")]

    split_content_list = [x.split("/") for x in content_list]
    
    print(content_list)
    print(split_content_list)

    for i, item in enumerate(split_content_list):
        if item[1] not in content_dict:
            content_dict[item[1]] = {}
        if item[2] not in content_dict[item[1]]:
            content_dict[item[1]][item[2]] = []
        content_dict[item[1]][item[2]].append({
            "name": item[3],
            "path": content_list[i]
            })
        
    for y in tqdm(years_images_to_extract):
        # get all the images for that year
        year_images = content_dict[str(y)]
        for m,v in tqdm(year_images.items(),leave=False):
            if m not in ["01","02","03","04","05","06","07","08","09","10","11","12"]:
                continue
            # print(k,v)
            for image in tqdm(v, leave=False):
                # pass
                # print(image)
                src = image["path"]
                
                bn_split = image["name"].split("-")
                ext = image["name"].split(".")[-1]
                new_bn = bn_split[0] + "." + ext
                tgt = f"{output_dir}/{y}/{m}/{new_bn}"

                os.makedirs(os.path.dirname(tgt), exist_ok=True)
                if os.path.exists(tgt):
                    continue
                
                with zip.open(src) as s, open(tgt, 'wb') as t:
                    shutil.copyfileobj(s, t)
            # break
        # break


# print(content_dict)
```

```python
import glob
drawings_png = sorted(glob.glob(f"{output_dir}/**/*.png", recursive=True))
drawings_jpg = sorted(glob.glob(f"{output_dir}/**/*.jpg", recursive=True))


def correct_filename_format (filelist):
    for i in tqdm(range(len(filelist))):
        src = filelist[i]
        src_bn = os.path.basename(src)

        if '-' in src_bn:
            bn_split = src_bn.split("-")
            ext = src_bn.split(".")[-1]

            new_bn = bn_split[0] + "." + ext
            tgt = os.path.dirname(src) + "/" + new_bn
            
            # print(src, "n", tgt)
            os.rename(src, tgt)

correct_filename_format(drawings_png)
correct_filename_format(drawings_jpg)
        
```

```python
def create_excentricity_map(mask, sun_radius, increasing=True, value_outside=np.nan, normalize=False):
    """
    Create a map of the excentricity of the sun in the image.
    args:
        mask: 2D array of the same size as the image
        sun_radius: radius of the sun in pixels
        increasing: if True, the excentricity map will be increasing from the center of the sun to the edge
                    if False, the excentricity map will be decreasing from the center of the sun to the edge
        value_outside: value to put outside the sun
        normalize: if True, the excentricity map will be normalized to be between 0 and 1
    """
    # get the center of the sun
    center = np.array(mask.shape)//2
    # create a meshgrid of the same size as the image
    x, y = np.meshgrid(np.arange(mask.shape[0]), np.arange(mask.shape[1]))
    # compute the distance between each pixel and the center of the sun
    dist = np.sqrt((x-center[0])**2 + (y-center[1])**2)

    ratio = dist/sun_radius
    ratio = np.clip(ratio, -1, 1)
    rho = np.arcsin(ratio)
    rho_deg = np.rad2deg(rho)

    rho_deg[rho_deg>=90] = value_outside
    
    if not increasing:
        # make the excentricity map decreasing 
        # (the center of the sun is 90 degrees, the edge is 0 degrees)
        rho_deg[rho_deg>=0] = 90 - rho_deg[rho_deg>=0]

    if normalize:
        # normalize the excentricity map to be between 0 and 1
        # (the center of the sun is 0, the edge is 1)
        # do not normalize the values outside the sun
        rho_deg[rho_deg>=0] = rho_deg[rho_deg>=0] / 90

    # compute the excentricity map
    excentricity_map = rho_deg

    return excentricity_map

def reframe_drawing( dr, calib, wl_rad, out_shape):
        out_factor = (out_shape[0] // 2) / wl_rad
        # print(f"out_factor: {out_factor}")
        sun_centre = calib['CenterX'], calib['CenterY']
        sun_radius = calib['Radius']*out_factor

        # frame is a bounding box around the sun with format [xmin, xmax, ymin, ymax] ou [left, right, bottom, top]
        sun_frame  = [sun_centre[0]-sun_radius, sun_centre[0]+sun_radius, 
                    sun_centre[1]-sun_radius, sun_centre[1]+sun_radius]
        
        # if any of the sun_frame values is negative or larger than the image size -> create a new image with the correct size
        if any([val < 0 for val in sun_frame]) or any([val > dim for val, dim in zip(sun_frame, dr.shape[:-1])]):
            top = int(-sun_frame[2]) if sun_frame[2] < 0 else 0
            bottom = int(sun_frame[3]-dr.shape[0]) if sun_frame[3] > dr.shape[0] else 0
            left = int(-sun_frame[0]) if sun_frame[0] < 0 else 0
            right = int(sun_frame[1]-dr.shape[1]) if sun_frame[1] > dr.shape[1] else 0

            # get mean value of the image
            mean_val = np.median(dr)

            dr_clip = cv2.copyMakeBorder(dr, top, bottom, left, right, cv2.BORDER_CONSTANT, value=mean_val)

            dr_clip = dr_clip[left+int(sun_frame[0]):left+int(sun_frame[1]),
                                top+int(sun_frame[2]):top+int(sun_frame[3])]
        else:
            dr_clip = dr[int(sun_frame[2]):int(sun_frame[3]), int(sun_frame[0]):int(sun_frame[1])]

        # To see the content as the DIRECT observation of the sun, flip the image horizontally (revert inversion due
        # to operator projection on the screen )
        dr_clip_flip = np.flip(dr_clip, axis=[1])


        # make it black and white and single channel
        dr_clip_flip = cv2.cvtColor(dr_clip_flip, cv2.COLOR_BGR2GRAY)
        resized_array = cv2.resize(dr_clip_flip, out_shape[::-1], interpolation=cv2.INTER_LINEAR)

        return resized_array

import math
def reframe_drawing2( dr, calib, wl_rad, out_shape, erosion=0):
        '''
        erosion: number of pixels to remove from the sun radius
        '''
        out_factor = (out_shape[0] // 2) / wl_rad
        # print(f"out_factor: {out_factor}")
        sun_centre = calib['CenterX'], calib['CenterY']
        sun_radius = calib['Radius']-erosion

        # frame is a bounding box around the sun with format [xmin, xmax, ymin, ymax] ou [left, right, bottom, top]
        sun_frame  = [sun_centre[0]-sun_radius, sun_centre[0]+sun_radius, 
                    sun_centre[1]-sun_radius, sun_centre[1]+sun_radius]
        dr_sun_only = dr[int(sun_frame[2]):int(sun_frame[3]), int(sun_frame[0]):int(sun_frame[1])]

        # expand the image to 2xwl_rad
        dr_sun_only_wl_size = cv2.resize(dr_sun_only, (2*math.ceil(wl_rad), 2*math.ceil(wl_rad)))

        # create new image with the correct size
        out_img = np.zeros((out_shape[0], out_shape[1], 3)).astype(np.uint8)

        # the add the drawing to the image
        out_centre = np.array(out_shape)//2
        out_img[out_centre[0] - dr_sun_only_wl_size.shape[0]//2:out_centre[0] + dr_sun_only_wl_size.shape[0]//2,
                out_centre[1] - dr_sun_only_wl_size.shape[1]//2:out_centre[1] + dr_sun_only_wl_size.shape[1]//2, :] = dr_sun_only_wl_size

        # fig2, ax2 = plt.subplots(1,4, figsize=(12,4))
        # ax2[0].imshow(dr)
        # ax2[1].imshow(dr_sun_only)
        # ax2[2].imshow(dr_sun_only_wl_size)
        # ax2[3].imshow(out_img)
        # fig2.show()
        
        # make it black and white and single channel
        out_img = cv2.cvtColor(out_img, cv2.COLOR_RGB2GRAY)

        # To see the content as the DIRECT observation of the sun, flip the image horizontally (revert inversion due
        # to operator projection on the screen )
        out_img = np.flip(out_img, axis=[1])

        return out_img
        
from PIL import Image

def refresh_correct_drawing(change):
        triplet = triplets_split[img_slider1.value]

        wl_fn = wl_dir / triplet["wl_name"]
        cal_fn = cal_dir / triplet["cal_name"]
        dr_fn = dr_dir / triplet["dr_name"]

        hdulst_wl:fits.HDUList = fits.open(wl_fn)
        img_wl = hdulst_wl[0].data
        header_wl = hdulst_wl[0].header

        hdulst_cal:fits.HDUList = fits.open(cal_fn)
        img_cal = hdulst_cal[0].data


        # img_dr = cv2.imread(str(dr_fn), cv2.IMREAD_GRAYSCALE)
        img_dr = Image.open(dr_fn).convert('RGB')
        img_dr = np.array(img_dr)
        # print(img_dr.shape)
        calib = triplet["dr_calib"]

        # print(calib)

        wl_rad = header_wl['SOLAR_R']

        framed_dr = reframe_drawing2(img_dr, calib, wl_rad, img_wl.shape, erosion=5)

        angle_map = np.zeros_like(framed_dr)
        angle_map = create_excentricity_map(angle_map, wl_rad, increasing=False , value_outside=-1, normalize=False) 
        sun_mask = angle_map.copy()
        # print(f"sun_mask.shape: {sun_mask.shape}, min: {sun_mask.min()}, max: {sun_mask.max()}")
        sun_mask[sun_mask>0] = 1
        sun_mask[sun_mask<0] = 0
        # print(f"sun_mask.shape: {sun_mask.shape}, min: {sun_mask.min()}, max: {sun_mask.max()}")


        ax[0].clear()
        ax[1].clear()
        ax[2].clear()
        ax[0].imshow(img_dr, cmap='gray')
        ax[1].imshow(framed_dr, cmap='gray')
        ax[1].imshow(sun_mask, cmap='jet', alpha=0.5)
        ax[2].imshow(framed_dr * sun_mask, cmap='gray')

        # add points to the drawing from the calib
        sun_centre = calib['CenterX'], calib['CenterY']
        ax[0].scatter(sun_centre[0], sun_centre[1], c='r', s=10)
        sun_north = calib['NorthX'], calib['NorthY']
        ax[0].scatter(sun_north[0], sun_north[1], c='r', s=10)
        sun_radius = calib['Radius']
        sun_compute_radius = np.sqrt((sun_centre[0]-sun_north[0])**2 + (sun_centre[1]-sun_north[1])**2)
        sun_east = sun_centre[0] + sun_radius, sun_centre[1]
        sun_west = sun_centre[0] - sun_radius, sun_centre[1]
        sun_south = sun_centre[0], sun_centre[1] + sun_radius

        ax[0].scatter(sun_east[0], sun_east[1], c='r', s=10)
        ax[0].scatter(sun_west[0], sun_west[1], c='r', s=10)
        ax[0].scatter(sun_south[0], sun_south[1], c='r', s=10)

        # print(f"sun_radius: {sun_radius}, sun_compute_radius: {sun_compute_radius}")
        


# split = "train"
split = "test"
triplets_split = train_triplets if split == "train" else test_triplets

last_index = 184
img_slider1 = widgets.IntSlider(min=0, max=len(triplets_split), step=1, value=last_index, description='Image Index')
img_slider1.observe(refresh_correct_drawing, names='value')


plt.ioff()
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
plt.ion()
refresh_correct_drawing(None)

# refresh_find_clip_values(None)
display(widgets.VBox([img_slider1,
                      fig.canvas]))
```

```python
def resize_cal(radius_wl, radius_cal, test_img_cal, test_img_wl, show=False):
   # get ratio between radii
   radius_ratio = radius_wl/radius_cal
   # print(radius_ratio)

   # resize cal image so that the sun has the same radius in both images
   # To resize, we need to know the new shape of the image
   # The new shape is the old one multiplied by the ratio of radii
   new_shape = np.array(test_img_cal.shape)*radius_ratio

   # use cv2 or numpy to resize the image with bilinear interpolation
   # cv2 is faster, but numpy is easier to use
   # interpolation = cv2.INTER_LINEAR
   interpolation = cv2.INTER_NEAREST

   resized = cv2.resize(test_img_cal.data, tuple(new_shape.astype(int)), interpolation=interpolation)

   # crop the resized image to match the shape of the wl image, padding with zeros

   resized_cropped = np.zeros_like(test_img_wl.data)
   # get the center of the resized image
   center_resized = np.array(resized.shape)//2
   # get the center of the cropped image
   center_cropped = np.array(resized_cropped.shape)//2
   # get the top left corner of the cropped image
   top_left = center_cropped - center_resized
   # get the bottom right corner of the cropped image
   bottom_right = top_left + resized.shape
   # crop the resized image
   resized_cropped[top_left[0]:bottom_right[0], top_left[1]:bottom_right[1]] = resized

   if show:
      print(resized.shape)
      # show the resized image
      plt.imshow(resized, cmap='gray')
      plt.show()
      print(resized_cropped.shape)
      # show the cropped image
      plt.imshow(resized_cropped, cmap='gray')
      plt.show()

   return resized_cropped

def resize_cal2(radius_wl, radius_cal, test_img_cal, test_img_wl, show=False):
   # get ratio between radii
   radius_ratio = radius_wl/radius_cal
   # print(radius_ratio)

   # 1) crop the cal image to the sun_only such that the sun is in the center of the image
   sun_centre_cal = np.array(test_img_cal.shape)//2
   sun_centre_wl = np.array(test_img_wl.shape)//2

   sun_frame_cal  = [sun_centre_cal[0]-radius_cal, sun_centre_cal[0]+radius_cal, 
                    sun_centre_cal[1]-radius_cal, sun_centre_cal[1]+radius_cal]
   cal_sun_only = test_img_cal.data[int(sun_frame_cal[2]):int(sun_frame_cal[3]), int(sun_frame_cal[0]):int(sun_frame_cal[1])]
   # print(f"cal_sun_only.shape: {cal_sun_only.shape}")

   sun_frame_wl  = [sun_centre_wl[0]-radius_wl, sun_centre_wl[0]+radius_wl, 
                    sun_centre_wl[1]-radius_wl, sun_centre_wl[1]+radius_wl]
   wl_sun_only = test_img_wl.data[int(sun_frame_wl[2]):int(sun_frame_wl[3]), int(sun_frame_wl[0]):int(sun_frame_wl[1])]
   # print(f"wl_sun_only.shape: {wl_sun_only.shape}")

   # 2) resize the cal_sun_only image to the size of the wl_sun_only image
   # use cv2 or numpy to resize the image with bilinear interpolation
   # make sure the new shape is a tuple of even numbers
   new_shape = np.array(wl_sun_only.shape)
   new_shape[new_shape%2==1] += 1
   resized_cal_sun_only = cv2.resize(cal_sun_only, new_shape, interpolation=cv2.INTER_LINEAR)
   # print(f"resize_cal_sun_only.shape: {resized_cal_sun_only.shape}")

   # 3) create a new image with the size of the wl image and put the resized cal_sun_only image in the center
   resized_cropped = np.zeros_like(test_img_wl.data)
   # get the center of the resized image
   center_resized = np.array(resized_cal_sun_only.shape)//2
   # get the center of the cropped image
   center_new_img = np.array(resized_cropped.shape)//2

   # print(f"centers: {center_resized}, {center_new_img}")

   resized_cropped[center_new_img[0]-center_resized[0]:center_new_img[0]+center_resized[0],
                     center_new_img[1]-center_resized[1]:center_new_img[1]+center_resized[1]] = resized_cal_sun_only

   
   if show:
      print(resized_cal_sun_only.shape)
      f = plt.figure(figsize=(12,4))
      f0 = f.add_subplot(1,4, 1)
      # show the original image
      f0.imshow(test_img_cal.data, cmap='gray')
      f1 = f.add_subplot(1,4, 2)
      # show the resized image
      f1.imshow(resized_cal_sun_only, cmap='gray')
      f2 = f.add_subplot(1,4, 3)
      print(resized_cropped.shape)
      # show the cropped image
      f2.imshow(resized_cropped, cmap='gray')
      f3 = f.add_subplot(1,4, 4)
      f3.imshow(test_img_wl.data, cmap='gray')
      plt.show()

   return resized_cropped


def refresh_correct_calcium(change):
   # save the current axis limits
   x0lim = ax2[0].get_xlim()
   y0lim = ax2[0].get_ylim()
   x1lim = ax2[1].get_xlim()
   y1lim = ax2[1].get_ylim()
   x2lim = ax2[2].get_xlim()
   y2lim = ax2[2].get_ylim()


   triplet = triplets_split[img_slider2.value]

   wl_fn = wl_dir / triplet["wl_name"]
   cal_fn = cal_dir / triplet["cal_name"]

   hdulst_wl:fits.HDUList = fits.open(wl_fn)
   img_wl = hdulst_wl[0].data
   header_wl = hdulst_wl[0].header
   radius_wl = header_wl['SOLAR_R']

   hdulst_cal:fits.HDUList = fits.open(cal_fn)
   img_cal = hdulst_cal[0].data
   header_cal = hdulst_cal[0].header
   center_cal = np.array(img_cal.shape)//2
   radius_cal = header_cal['SOLAR_R']


   resized_cal = resize_cal2(radius_wl, radius_cal, hdulst_cal[0], hdulst_wl[0], show=False)

   # clear all axes of the figure
   for axx in ax2:
      axx.clear()

   ax2[0].imshow(img_cal, cmap='gray')
   ax2[1].imshow(img_wl, cmap='gray')
   #overlay the calcium image on the white light image
   ax2[2].imshow(img_wl, cmap='gray')
   ax2[2].imshow(img_cal, cmap='Blues', alpha=0.5)

   if show_sun_mask.value:
      ax2[0].imshow(resized_cal, cmap='jet', alpha=0.25)
      ax2[1].imshow(resized_cal, cmap='jet', alpha=0.25)
      ax2[2].imshow(resized_cal, cmap='Reds', alpha=0.25)
   

   if not x0lim == (0,1):
      ax2[0].set_xlim(x0lim)
      ax2[0].set_ylim(y0lim)
      ax2[1].set_xlim(x1lim)
      ax2[1].set_ylim(y1lim)
      ax2[2].set_xlim(x2lim)
      ax2[2].set_ylim(y2lim)



split = "train"
# split = "test"
triplets_split = train_triplets if split == "train" else test_triplets

last_index = 0
img_slider2 = widgets.IntSlider(min=0, max=len(triplets_split), step=1, value=last_index, description='Image Index')
img_slider2.observe(refresh_correct_drawing, names='value')

# add a checkbox to overlat the new calcium image on the original calcium image
show_sun_mask = widgets.Checkbox(value=False, description='Show Resized Calcium')
show_sun_mask.observe(refresh_correct_calcium, names='value')


plt.ioff()
fig2, ax2 = plt.subplots(1, 3, figsize=(15, 5))
# fig2, ax2 = plt.subplots(2, 2, figsize=(10, 10))
plt.ion()
refresh_correct_calcium(None)

# refresh_find_clip_values(None)
display(widgets.VBox([img_slider2,
                        show_sun_mask,
                      fig2.canvas]))
```

```python
def crop_with_padding(arr, start_row, end_row, start_col, end_col, padding_value=0):
    original_shape = arr.shape
    
    # Calculate the cropping region
    crop_start_row = max(0, start_row)
    crop_end_row = min(original_shape[0], end_row)
    crop_start_col = max(0, start_col)
    crop_end_col = min(original_shape[1], end_col)
    
    # Calculate padding dimensions
    pad_top = max(0, -start_row)
    pad_bottom = max(0, end_row - original_shape[0])
    pad_left = max(0, -start_col)
    pad_right = max(0, end_col - original_shape[1])
    
    # Create the cropping region
    cropped = arr[crop_start_row:crop_end_row, crop_start_col:crop_end_col]
    
    # Pad the cropping region if needed
    padded = np.pad(cropped, ((pad_top, pad_bottom), (pad_left, pad_right)), mode='constant', constant_values=padding_value)
    
    return padded

from typing import Literal
NORM_TYPES = Literal['max', 'percentile']

def crop_triplets_active_regions_v2(triplet, crop_side, 
                                           get_wl=True, get_cal=True, get_dr=True, get_angle_map=True, 
                                           clip_off_disk=True, off_disk_wl=1000, off_disk_cal=1000,
                                           norm_wl=True, norm_wl_method: NORM_TYPES='max',
                                           norm_cal=True, norm_cal_method: NORM_TYPES='percentile',
                                           norm_angle_map=True,
                                           scale_between_0_255=False
                                           ):

    active_regions = triplet["active_regions"]
    
    wl_fn = wl_dir / triplet["wl_name"]
    cal_fn = cal_dir / triplet["cal_name"]
    dr_fn = dr_dir / triplet["dr_name"]


    #### FULL DISK PHASE ####
    if get_wl:
        #################### DUMP WL ####################
        hdulst_wl:fits.HDUList = fits.open(wl_fn)
        test_img_wl = hdulst_wl[0]
        header_wl = test_img_wl.header
        center_wl = np.array(test_img_wl.shape)//2
        radius_wl = header_wl['SOLAR_R']

        # create a mask with the same size as the image, and a circle in the middle with radius radius_wl
        mask = np.zeros_like(test_img_wl.data)    
        cv2.circle(mask, tuple(center_wl), int(radius_wl*0.99), 1, -1)

        wl_content = test_img_wl.data.copy()
        # To see the content as the DIRECT observation of the sun, flip the image vertically ( revert sunpy's flip)
        wl_content = np.flip(wl_content, axis=0)

        if clip_off_disk:
            wl_content[np.where(wl_content < off_disk_wl )] = 0

        if norm_wl:
            if norm_wl_method == 'max':
                wl_content = wl_content / np.max(wl_content)
            elif norm_wl_method == 'percentile':
                wl_content = wl_content / np.percentile(wl_content, 99.9)

        if scale_between_0_255:
            wl_content = scale_to_255(wl_content)
            


    if get_cal:
        # get cal image
        hdulst_cal:fits.HDUList = fits.open(cal_fn)
        test_img_cal = hdulst_cal[0]
        header_cal = test_img_cal.header
        center_cal = np.array(test_img_cal.shape)//2
        radius_cal = header_cal['SOLAR_R']


        # resized_cropped_cal = resize_cal(radius_wl, radius_cal, test_img_cal, test_img_wl, show=False)
        resized_cropped_cal = resize_cal2(radius_wl, radius_cal, test_img_cal, test_img_wl, show=False)

        cal_content = resized_cropped_cal.copy()
        # To see the content as the DIRECT observation of the sun, flip the image vertically ( revert sunpy's flip)
        cal_content = np.flip(cal_content, axis=0)

        if clip_off_disk:
            cal_content[np.where(cal_content < off_disk_cal )] = 0

        if norm_cal:
            if norm_cal_method == 'max':
                cal_content = cal_content / np.max(cal_content)
            elif norm_cal_method == 'percentile':
                cal_content = cal_content / np.percentile(cal_content, 99.9)

        if scale_between_0_255:
            cal_content = scale_to_255(cal_content)
    
    if get_dr:
        # dr = plt.imread(dr_fn)
        dr = Image.open(dr_fn).convert("RGB")  # Some images are 4-channel RGBA or CMYK
                                               # Convert them to standard 3-channel RGB
        dr = np.array(dr)
        calib = triplet["dr_calib"]
        wl_rad = header_wl['SOLAR_R']

        # resized_array = reframe_drawing(dr, calib, wl_rad, img_wl.shape)
        resized_array = reframe_drawing2(dr, calib, wl_rad, test_img_wl.shape, erosion=5)

        # create a mask with the same size as the dr_clip_flip, and a sphere in the middle with radius radius_wl
        # -1 is the background, positive values are the sphere such that the value is the andular distance to the center
        mask = np.zeros_like(resized_array)
        angle_map = create_excentricity_map(mask, wl_rad, increasing=False , value_outside=-1, normalize=norm_angle_map)  
        # -1 off disk, 0 on edge of disk, 1 center
        
        sun_mask = angle_map.copy()
        sun_mask[sun_mask>0] = 1
        sun_mask[sun_mask<0] = 0


        if clip_off_disk:
            resized_array = resized_array * sun_mask

        # NOT NEEDED TO NORMALIZE THE DRAWING / ALREADY IN JPEG    
        # if normalize:
        #     # normalize the image
        #     resized_array = resized_array / np.max(resized_array)

        if scale_between_0_255:
            angle_map = scale_to_255(angle_map)

        
    #### CROP PHASE ####

    crops_wl = []
    crops_cal = []
    crops_dr = []
    crops_angle_map = []


    for i, active_region in enumerate(active_regions):
            
        if get_wl:
            # get the crop of the active region
            obstime = header_wl["DATE-OBS"]
            obstime = obstime.replace(" ", "T")

            header_wl.append(('CTYPE1', 'HPLN-TAN'))
            header_wl.append(('CTYPE2', 'HPLT-TAN'))
            header_wl["DATE-OBS"] = obstime
            header_wl["MJD-OBS"] = Time(obstime).mjd

            wcs_wl = WCS(header_wl)

            cur_sk_wl = SkyCoord(active_region["Longitude"]*u.rad, active_region["Latitude"]*u.rad , frame=frames.HeliographicCarrington,
                            obstime=obstime, observer="earth") 
            coords_wl = skycoord_to_pixel(cur_sk_wl, wcs_wl, origin=0)

            # Get the coords in the direct observation (vertical flip of the other)
            coords_wl_direct = np.array([coords_wl[0], test_img_wl.data.shape[1]-coords_wl[1]])

            coords_wl_direct = np.array(coords_wl_direct, dtype=int)
            crop_wl = crop_with_padding(wl_content,
                                        coords_wl_direct[1]-crop_side//2 , coords_wl_direct[1]+crop_side//2,
                                        coords_wl_direct[0]-crop_side//2 , coords_wl_direct[0]+crop_side//2,
                                        padding_value=-0)
            
            crops_wl.append((crop_wl, header_wl['SIMPLE'], header_wl['BITPIX'],
                                 header_wl['NAXIS1'], header_wl['NAXIS2'], 
                                 header_wl['DATE-OBS'], header_wl['MJD-OBS']))
            
        if get_cal:

            # get the crop of the active region
            obstime = header_wl["DATE-OBS"]
            obstime = obstime.replace(" ", "T")
            
            header_cal.append(('CTYPE1', 'HPLN-TAN'))
            header_cal.append(('CTYPE2', 'HPLT-TAN'))
            header_cal["DATE-OBS"] = obstime
            header_cal["MJD-OBS"] = Time(obstime).mjd

            coords_cal_direct = np.array(coords_wl_direct, dtype=int)

            crop_cal = crop_with_padding(cal_content,
                                        coords_cal_direct[1]-crop_side//2 , coords_cal_direct[1]+crop_side//2,
                                        coords_cal_direct[0]-crop_side//2 , coords_cal_direct[0]+crop_side//2,
                                        padding_value=-0)
            

            crops_cal.append((crop_cal, header_cal['SIMPLE'], header_cal['BITPIX'],
                                 header_cal['NAXIS1'], header_cal['NAXIS2'], 
                                 header_cal['DATE-OBS'], header_cal['MJD-OBS']))

        if get_dr:
            
            crop_dr = crop_with_padding(resized_array, 
                            coords_wl_direct[1]-crop_side//2 , coords_wl_direct[1]+crop_side//2,
                            coords_wl_direct[0]-crop_side//2, coords_wl_direct[0]+crop_side//2,
                            padding_value=-0)
            

            crops_dr.append(crop_dr)

            if get_angle_map:
                crop_angle_map = crop_with_padding(angle_map, 
                                coords_wl_direct[1]-crop_side//2 , coords_wl_direct[1]+crop_side//2,
                                coords_wl_direct[0]-crop_side//2, coords_wl_direct[0]+crop_side//2,
                                padding_value=-1)
                crops_angle_map.append(crop_angle_map)




    return crops_wl, crops_cal, crops_dr, crops_angle_map
```

```python
import cv2
def rotate_given_angle(data, angle=0):
    # we consider rotating the wl on the cal
    rotated_data = data.copy()
    (h, w) = rotated_data.shape[:2]
    (cX, cY) = (w // 2, h // 2)
    # rotate our image by 45 degrees around the center of the image
    M = cv2.getRotationMatrix2D((cX, cY), angle, 1.0)
    rotated_data = cv2.warpAffine(rotated_data, M, (w, h))

    return rotated_data

def LoG_filter_opencv(image, sigma, size=None):
    # Generate LoG kernel
    if size is None:
        size = int(6 * sigma + 1) if sigma >= 1 else 7

    if size % 2 == 0:
        size += 1

    x, y = np.meshgrid(np.arange(-size//2+1, size//2+1), np.arange(-size//2+1, size//2+1))
    kernel = -(1/(np.pi * sigma**4)) * (1 - ((x**2 + y**2) / (2 * sigma**2))) * np.exp(-(x**2 + y**2) / (2 * sigma**2))
    kernel = kernel / np.sum(np.abs(kernel))

    # Perform convolution using OpenCV filter2D
    result = cv2.filter2D(image, -1, kernel)

    return result
```

```python
from pathlib import Path 
import os
import glob

# function that gets all the files in a directory

def get_files_in_dir(dir_path, full_path=True, ext="npz"):
    files = sorted(glob.glob(str(dir_path / f"*.{ext}")))
    
    if full_path:
        return files
    else:
        return [os.path.basename(file) for file in files]
    

split = "train"
# split = "test"

save_to_png = False
ext = "png" if save_to_png else "npz"

sub_dir = "new_crops"

wl_out_dir = Path(f"../datasets/ImageTranslationDEC2023/{split}/{sub_dir}/whitelight/")
cal_out_dir = Path(f"../datasets/ImageTranslationDEC2023/{split}/{sub_dir}/calcium/")
dr_out_dir = Path(f"../datasets/ImageTranslationDEC2023/{split}/{sub_dir}/drawing/")
angle_map_out_dir = Path(f"../datasets/ImageTranslationDEC2023/{split}/{sub_dir}/dr_angle_map/")

wl_npz_fn = get_files_in_dir(wl_out_dir, full_path=False, ext=ext)
cal_npz_fn = get_files_in_dir(cal_out_dir, full_path=False, ext=ext)
dr_npz_fn = get_files_in_dir(dr_out_dir, full_path=False, ext=ext)
angle_map_npz_fn = get_files_in_dir(angle_map_out_dir, full_path=False, ext=ext)

# write each list in a txt file
outfile_wl = f"../datasets/ImageTranslationDEC2023/{split}_{sub_dir}_whitelight.txt"
outfile_cal = f"../datasets/ImageTranslationDEC2023/{split}_{sub_dir}_calcium.txt"
outfile_dr = f"../datasets/ImageTranslationDEC2023/{split}_{sub_dir}_drawing.txt"
outfile_angle_map = f"../datasets/ImageTranslationDEC2023/{split}_{sub_dir}_dr_angle_map.txt"

with open(outfile_wl, "w") as f:
    for item in wl_npz_fn:
        item = f"{split}/{sub_dir}/whitelight/{item}"
        f.write("%s\n" % item)

with open(outfile_cal, "w") as f:
    for item in cal_npz_fn:
        item = f"{split}/{sub_dir}/calcium/{item}"
        f.write("%s\n" % item)
        
with open(outfile_dr, "w") as f:
    for item in dr_npz_fn:
        item = f"{split}/{sub_dir}/drawing/{item}"
        f.write("%s\n" % item)

with open(outfile_angle_map, "w") as f:
    for item in angle_map_npz_fn:
        item = f"{split}/{sub_dir}/dr_angle_map/{item}"
        f.write("%s\n" % item)

```
